## Objective
To compare the performance of traditional machine learning

## 1. Models Compared

The following models are evaluated:

### Linear Regression (Baseline)
- Treats each observation independently
- Simple and interpretable
- Serves as a performance benchmark

### LSTM (Sequence-Based Model)
- Learns temporal dependencies
- Uses sliding window sequences
- Designed for time-series forecasting


## 2. Evaluation Metrics

The following metrics are used for comparison:

- **MAE (Mean Absolute Error)**  
  Measures average absolute prediction error

- **RMSE (Root Mean Squared Error)**  
  Penalizes large prediction errors

- **R² Score**  
  Measures variance explained by the model

## 4. Qualitative Model Comparison

### Linear Regression
- Captures overall trend
- Struggles with sudden consumption changes
- Cannot model temporal dependencies

### LSTM
- Learns sequential patterns
- Shows smoother predictions
- Better suited for time-dependent energy usage


## 5. Final Model Selection

While Linear Regression provides a strong and interpretable baseline,
it fails to capture temporal dependencies inherent in energy data.

The LSTM model, despite evaluation constraints,
demonstrates superior ability to learn time-based patterns.

### Selected Model for Production:
**LSTM**

### Reason:
- Designed for sequential data
- Captures daily consumption cycles
- Scalable to longer horizons and additional features


## 6. Limitations and Future Improvements

- Limited test sequences after rolling feature creation
- External factors (weather, occupancy) not included
- Model performance can improve with:
  - More historical data
  - Device-level features
  - Hybrid models (LSTM + regression)
